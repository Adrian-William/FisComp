```python
import numpy as np
import matplotlib.pyplot as plt
import os
import random as rd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import sequential
from tensorflow.keras.layers import Dense, Dropout


# Load the data from the text files
pesos = np.loadtxt('pesos.txt')
bias = np.loadtxt('bias.txt')

if pesos == 0  or bias == 0:
    print("Gerando novos pesos e bias")
    # Gerando novos pesos e bias
    pesos = {camada1 = np.random.rand(10),
             camada2 =  np.random.rand(10),
             camada3 =  np.random.rand(10)}
    bias = {camada1 =  np.random.rand(10),
            camada2 =  np.random.rand(10),
            camada3 = np.random.rand(10)}

def backpropagation(X, y, pesos, bias, learning_rate=0.01, epochs=1000):
    for epoch in range(epochs):
        for i in range(len(X)):
            
            # Forward pass
            passo1 = np.dot(X[i], pesos['camada1']) + bias['camada1']
            passo2 = np.dot(passo1, pesos['camada2']) + bias['camada2']
            output = np.dot(passo2, pesos['camada3']) + bias['camada3']
            # Compute the error
            error = y[i] - output
            # Backward pass
            d_output = error * output * (1 - output)  # Derivative of sigmoid
            d_passo2 = np.dot(d_output, pesos['camada3'].T) * passo2 * (1 - passo2)
            d_passo1 = np.dot(d_passo2, pesos['camada2'].T) * passo1 * (1 - passo1)
            # Update weights and biases
            pesos['camada3'] += np.dot(passo2.T, d_output) * learning_rate
            bias['camada3'] += np.sum(d_output, axis=0) * learning_rate
            pesos['camada2'] += np.dot(passo1.T, d_passo2) * learning_rate
            bias['camada2'] += np.sum(d_passo2, axis=0) * learning_rate
            pesos['camada1'] += np.dot(X[i].T, d_passo1) * learning_rate
            bias['camada1'] += np.sum(d_passo1, axis=0) * learning_rate
    return pesos, bias

```
