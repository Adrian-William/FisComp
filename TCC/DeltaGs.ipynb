```python
import matplotlib.pyplot as plt
import numpy as np

def plot_diagrama_rro(En_O2H, En_O, En_OH, U_cal, Dopagem):


    # 1. Defina os passos da reação e as energias livres (Delta G) calculadas
    passos = ['O2', 'OOH*', 'O*', 'OH*', 'H2O']

    # Exemplo de valores de energia (substitua pelos SEUS resultados)
    # Esses valores são apenas exemplos ilustrativos para um caminho ladeira abaixo (U = 0V)
    # 1. Definição correta dos seus valores de Delta G (sem inverter os sinais)
    Delta_G = [En_O2H, En_O, En_OH]  
    Eletrons_livres = np.array([4.0, 3.0, 2.0, 1.0, 0.0]) 

    energias_U_0V = np.array([4.92, Delta_G[0], Delta_G[1], Delta_G[2], 0.00])
    energias_U_1_23V = energias_U_0V - (1.23 * Eletrons_livres)
    energias_U_cal = energias_U_0V - (U_cal * Eletrons_livres)

    # Configurações do gráfico
    fig, ax = plt.subplots(figsize=(8, 6))

    # Largura de cada linha horizontal (nível de energia)
    largura_linha = 0.6

    # 2. Loop para desenhar as linhas e as conexões
    for i in range(len(passos)):
        # Ponto x central do passo atual
        x_centro = i
        
        # Desenha a linha horizontal grossa para o nível de energia
        ax.hlines(y=energias_U_0V[i], xmin=x_centro - largura_linha/2, xmax=x_centro + largura_linha/2, 
                color='blue', linewidth=3, label='U = 0 V' if i == 0 else "")
        
        # Desenha a linha tracejada conectando ao próximo passo (se não for o último)
        if i < len(passos) - 1:
            # Ponto inicial do tracejado (fim da linha atual)
            x_inicio_tracejado = x_centro + largura_linha/2
            y_inicio_tracejado = energias_U_0V[i]
            
            # Ponto final do tracejado (início da próxima linha)
            x_fim_tracejado = (i + 1) - largura_linha/2
            y_fim_tracejado = energias_U_0V[i + 1]
            
            ax.plot([x_inicio_tracejado, x_fim_tracejado], 
                    [y_inicio_tracejado, y_fim_tracejado], 
                    color='blue', linestyle='--', linewidth=1.5)
            
    for i in range(len(passos)):
        # Ponto x central do passo atual
        x_centro = i
        
        # Desenha a linha horizontal grossa para o nível de energia
        ax.hlines(y=energias_U_1_23V[i], xmin=x_centro - largura_linha/2, xmax=x_centro + largura_linha/2, 
                color='green', linewidth=3, label='U = 1.23 V' if i == 0 else "")
        
        # Desenha a linha tracejada conectando ao próximo passo (se não for o último)
        if i < len(passos) - 1:
            # Ponto inicial do tracejado (fim da linha atual)
            x_inicio_tracejado = x_centro + largura_linha/2
            y_inicio_tracejado = energias_U_1_23V[i]
            
            # Ponto final do tracejado (início da próxima linha)
            x_fim_tracejado = (i + 1) - largura_linha/2
            y_fim_tracejado = energias_U_1_23V[i + 1]
            
            ax.plot([x_inicio_tracejado, x_fim_tracejado], 
                    [y_inicio_tracejado, y_fim_tracejado], 
                    color='green', linestyle='--', linewidth=1.5)
    for i in range(len(passos)):
        # Ponto x central do passo atual
        x_centro = i
        
        # Desenha a linha horizontal grossa para o nível de energia
        ax.hlines(y=energias_U_cal[i], xmin=x_centro - largura_linha/2, xmax=x_centro + largura_linha/2, 
                color='red', linewidth=3, label='U = -0.21 V' if i == 0 else "")
        
        # Desenha a linha tracejada conectando ao próximo passo (se não for o último)
        if i < len(passos) - 1:
            # Ponto inicial do tracejado (fim da linha atual)
            x_inicio_tracejado = x_centro + largura_linha/2
            y_inicio_tracejado = energias_U_cal[i]
            
            # Ponto final do tracejado (início da próxima linha)
            x_fim_tracejado = (i + 1) - largura_linha/2
            y_fim_tracejado = energias_U_cal[i + 1]
            
            ax.plot([x_inicio_tracejado, x_fim_tracejado], 
                    [y_inicio_tracejado, y_fim_tracejado], 
                    color='red', linestyle='--', linewidth=1.5)


    # 3. Formatação dos Eixos
    ax.set_xticks(range(len(passos)))
    ax.set_xticklabels(passos, fontsize=14)
    ax.set_ylabel('Free Energy (eV)', fontsize=14)
    ax.tick_params(axis='y', labelsize=12)

    # Adiciona uma legenda
    ax.legend(fontsize=12)

    # Adiciona título (Opcional)
    ax.set_title('Reaction free energy diagram for ' + Dopagem, fontsize=16)

    # Salva a imagem
    plt.tight_layout()
    plt.savefig('diagrama_rro.png', dpi=300)
    plt.show()
```
