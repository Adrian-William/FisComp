# Franks Hertz

```python
import matplotlib.pyplot as plt
import numpy as np
import os

file_path = r"medidas\medida_160C_10v_45v_01uv.msr"

if not os.path.exists(file_path):
    print("Arquivo não encontrado:", file_path)
else:
    with open(file_path, "rb") as f:
        data = f.read()

    # Extrai payload após cabeçalho conhecido
    payload_offset = 34
    payload = data[payload_offset:]

    # Interpreta como float32 (mais comum para dados científicos)
    arr = np.frombuffer(payload, dtype=np.float32)
    
    # Gera eixo x (tensão) de 10V a 45V (baseado no nome do arquivo)
    x = np.linspace(10, 45, len(arr))
    
    # Plotagem científica
    plt.figure(figsize=(10,6))
    plt.plot(x, arr, 'b-', linewidth=1.5, label='Dados experimentais')
    plt.grid(True, alpha=0.3)
    plt.xlabel('Tensão (V)')
    plt.ylabel('Corrente (unidades arb.)')
    plt.title('Experimento de Franck-Hertz (T = 160°C)')
    plt.legend()
    plt.show()

    # Info útil
    print(f"Número de pontos: {len(arr)}")
    print(f"Range de valores: [{arr.min():.2e}, {arr.max():.2e}]")
```
